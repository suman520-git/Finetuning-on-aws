```python
import sagemaker
from sagemaker.huggingface import HuggingFaceModel

role = sagemaker.get_execution_role()
```

```python
role
```

```python
model = HuggingFaceModel(
    model_data="s3://llm-model-artifacts-suman/models/huggingface-pytorch-training-2026-02-16-15-32-30-037/output/model.tar.gz",
    role=role,
    entry_point="inference.py",
    source_dir="./inference",
    transformers_version="4.37.0",
    pytorch_version="2.1.0",
    py_version="py310",
)
```

```python
predictor = model.deploy(
    initial_instance_count=1,
    instance_type="ml.g5.xlarge",
    endpoint_name="live-finetune-my-endpoint"
)
```

```python
## THIS IS JUST TO VALIDATE WHETHER MODEL WORKING OR NOT IN THE NOTEBOOK ITSELF
predictor.predict({"inputs": "Explain AWS S3"})
```
